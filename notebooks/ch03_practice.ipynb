## 62. 반복 점검 함수

```python
def inspect_dataframe(
    name: str,
    df: pd.DataFrame,
    primary_key: str | None = None,
) -> dict:
    result = {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": df.columns.tolist(),
        "dtypes": {column: str(dtype) for column, dtype in df.dtypes.items()},
        "missing_total": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
    }
    if primary_key is not None:
        result["primary_key"] = primary_key
        result["key_missing"] = int(df[primary_key].isna().sum())
        result["key_duplicates"] = int(df[primary_key].duplicated().sum())
        result["key_unique"] = int(df[primary_key].nunique())
    return result
```

```python
inspection_results = []
for name, df in datasets.items():
    inspection_results.append(
        inspect_dataframe(
            name,
            df,
            primary_keys[name],
        )
    )
inspection_results
```

> 원본 행이 아닌 구조 점검 요약을 기준으로 작성했습니다. 실제 값의 범위·분포가 필요한 항목은 추가 확인 대상으로 구분했습니다.

## 1. 데이터셋 및 컬럼·타입

### 데이터셋 구조

| 데이터셋          | 한 행의 의미          |   행 |  열 | 기본 키            | 외래 키                     |
| ------------- | ---------------- | --: | -: | --------------- | ------------------------ |
| `customers`   | 고객 1명            | 150 |  6 | `customer_id`   | 없음                       |
| `products`    | 상품 1개            | 100 |  4 | `product_id`    | 없음                       |
| `orders`      | 주문 1건            | 301 |  5 | `order_id`      | `customer_id`            |
| `order_items` | 주문에 포함된 상품 항목 1건 | 765 |  5 | `order_item_id` | `order_id`, `product_id` |

`orders`는 주문 단위이고, `order_items`는 주문에 포함된 상품 항목 단위입니다. 따라서 한 주문에 여러 상품이 포함되면 `order_items.order_id`가 여러 행에서 반복되는 것은 정상입니다.

### 컬럼과 타입

| 데이터셋          | 정수형 `int64`                                                         | 문자열형 `str`                                     | 주요 확인 사항                                    |
| ------------- | ------------------------------------------------------------------- | ---------------------------------------------- | ------------------------------------------- |
| `customers`   | `customer_id`, `age`                                                | `name`, `gender`, `city`, `signup_date`        | `customer_id`는 식별자, `signup_date`는 날짜 변환 필요 |
| `products`    | `product_id`, `price`                                               | `product_name`, `category`                     | `product_id`는 식별자, `price`는 범위 확인 필요        |
| `orders`      | `order_id`, `customer_id`                                           | `order_date`, `payment_method`, `order_status` | ID는 연결 키, `order_date`는 날짜 변환 필요            |
| `order_items` | `order_item_id`, `order_id`, `product_id`, `quantity`, `unit_price` | 없음                                             | ID와 수량·단가를 구분해서 해석해야 함                      |

ID 컬럼은 정수형이지만 계산용 숫자가 아니라 행을 식별하거나 파일을 연결하는 키입니다. 따라서 ID의 평균이나 합계에는 일반적인 분석 의미를 부여하지 않습니다.

| 추가 점검  | 숫자형 실제 범위 미확인                                 | 음수·0·극단값 확인 필요                    |
| 추가 점검  | 범주형 실제 값 미확인                                  | 오탈자·공백·표기 혼재 확인 필요                |
| 추가 점검  | `products.price`와 `order_items.unit_price` 관계 | 현재 가격과 주문 당시 단가일 수 있어 차이 원인 확인 필요 |
| 추가 점검  | 주문 상세가 없는 주문                                  | 반대 방향의 연결 여부 확인 필요                |
| 추가 점검  | 동일 주문·상품 조합 반복                                | 정상 분할 행인지 중복 적재인지 확인 필요           |

기본 키 중복, 전체 행 중복, 일반 결측치는 발견되지 않았습니다. 현재 가장 명확한 문제는 세 외래 키 관계의 참조 불일치입니다.

## 4. 즉시 처리하지 않은 이유

| 대상                       | 즉시 삭제·수정하지 않는 이유                                         |
| ------------------------ | -------------------------------------------------------- |
| 외래 키 불일치 행               | 추출 시점 차이, 기준정보 누락, 과거 기록, ID 입력 오류, 테스트 데이터 등 여러 원인이 가능함 |
| 반복되는 외래 키                | 한 고객의 여러 주문, 한 주문의 여러 상품처럼 정상적인 관계에서 발생할 수 있음            |
| 동일 고객 이름·상품명             | 서로 다른 고객이나 상품이 같은 이름을 사용할 수 있음                           |
| 0원 가격·단가                 | 무료 상품, 증정품 또는 프로모션일 가능성이 있음                              |
| `price`와 `unit_price` 차이 | 현재 상품 가격과 주문 당시 판매 단가가 다를 수 있음                           |
| 동일 주문·상품 조합              | 같은 상품을 별도 항목으로 기록했을 수도 있어 자동 중복 삭제가 위험함                  |
| 날짜 순서 이상                 | 데이터 이관이나 가입 정보 기준 차이 등 업무상 원인이 있을 수 있음                   |

따라서 이상 후보는 바로 제거하지 않고 별도 검토 대상으로 분리한 후 원인에 따라 수정·보완·제외 여부를 결정하는 것이 안전합니다.

## 종합 의견

현재 데이터는 **기본 키, 일반 결측치, 전체 중복 측면에서는 전반적으로 양호**합니다. 가장 먼저 확인할 사항은 세 외래 키 관계에서 각각 발견된 참조 불일치 1건이며, 실제 고유 문제 행 수와 발생 원인을 확인한 후 처리해야 합니다.

그다음 날짜 변환, 숫자형 범위, 범주형 표기, 가격 컬럼의 의미를 점검하고, 마지막으로 테이블 병합과 집계 기준을 확정하는 순서가 적절합니다.
