## 62. 반복 점검 함수

```python
def inspect_dataframe(
    name: str,
    df: pd.DataFrame,
    primary_key: str | None = None,
) -> dict:
    result = {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": df.columns.tolist(),
        "dtypes": {column: str(dtype) for column, dtype in df.dtypes.items()},
        "missing_total": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
    }
    if primary_key is not None:
        result["primary_key"] = primary_key
        result["key_missing"] = int(df[primary_key].isna().sum())
        result["key_duplicates"] = int(df[primary_key].duplicated().sum())
        result["key_unique"] = int(df[primary_key].nunique())
    return result
```

```python
inspection_results = []
for name, df in datasets.items():
    inspection_results.append(
        inspect_dataframe(
            name,
            df,
            primary_keys[name],
        )
    )
inspection_results
```

다만 특정 컬럼의 값이 반복되는 것과 전체 행 중복은 구분해야 합니다. 다음 반복은 업무 구조상 정상일 수 있습니다.

## 6. 숫자형 범위

구조 요약에는 숫자형 컬럼의 최솟값, 최댓값, 중앙값, 사분위수 등이 포함되어 있지 않아 실제 범위는 확인할 수 없습니다.

### 범위 점검 대상

| 데이터셋          | 컬럼           | 확인할 내용                              |
| ------------- | ------------ | ----------------------------------- |
| `customers`   | `age`        | 0 이하, 비현실적으로 큰 값, 특정 값에 과도하게 집중된 현상 |
| `products`    | `price`      | 음수 가격, 0원 가격, 지나치게 큰 값              |
| `order_items` | `quantity`   | 0 이하 수량, 지나치게 큰 주문 수량               |
| `order_items` | `unit_price` | 음수 단가, 0원 단가, 지나치게 큰 값              |

`customer_id`, `product_id`, `order_id`, `order_item_id`는 정수형이지만 식별자이므로 숫자형 범위 분석 대상에서 제외하는 것이 적절합니다.

`price`와 `unit_price`의 0도 즉시 오류로 단정해서는 안 됩니다. 실제 업무에서 허용되는 값인지 확인한 후 판단해야 합니다.

**pandas 확인 목적:**
분석 대상 숫자형 컬럼의 최솟값, 최댓값, 분위수와 빈도 분포를 확인하여 비정상 범위와 이상값 후보를 찾습니다.

## 7. 범주형 값

구조 요약에는 범주형 컬럼의 실제 고유값과 빈도가 포함되어 있지 않습니다.

### 점검 대상

| 데이터셋        | 컬럼               | 확인할 내용                      |
| ----------- | ---------------- | --------------------------- |
| `customers` | `gender`         | 표기 종류, 공백, 대소문자 또는 표현 방식 혼재 |
| `customers` | `city`           | 도시명 표기 일관성, 앞뒤 공백           |
| `products`  | `category`       | 카테고리 오탈자와 유사 카테고리 중복        |
| `orders`    | `payment_method` | 결제수단 표기 방식의 일관성             |
| `orders`    | `order_status`   | 주문상태 종류와 표기 일관성             |

실제 값이 제공되지 않았으므로 어떤 값이 정상이고 어떤 값이 오류인지는 현재 판단할 수 없습니다.

예를 들어 비슷한 표현이 여러 개 발견되더라도 실제로 서로 다른 상태인지 먼저 업무 의미를 확인해야 합니다.

**pandas 확인 목적:**
각 범주형 컬럼의 고유값, 값별 건수, 앞뒤 공백 여부를 확인하여 오탈자와 표기 불일치 후보를 찾습니다.

## 8. 날짜 범위와 변환 실패

날짜 관련 컬럼은 다음과 같습니다.

| 데이터셋        | 컬럼            | 현재 타입 |
| ----------- | ------------- | ----- |
| `customers` | `signup_date` | `str` |
| `orders`    | `order_date`  | `str` |

두 날짜 컬럼 모두 문자열로 저장되어 있습니다. 구조 요약에는 실제 날짜의 시작일, 종료일, 변환 실패 건수가 포함되어 있지 않으므로 날짜 범위는 확인할 수 없습니다.

### 추가 확인 사항

* 날짜형으로 변환되지 않는 값이 있는지
* 여러 날짜 형식이 혼재하는지
* 존재하지 않는 날짜가 있는지
* 날짜 앞뒤에 공백이 있는지
* 가입일과 주문일의 범위가 분석 목적에 적절한지
* 주문일이 고객 가입일보다 앞선 행이 있는지

마지막 항목은 두 데이터셋에 실제로 존재하는 `customer_id`, `signup_date`, `order_date`를 연결해 확인할 수 있습니다. 다만 해당 관계가 반드시 오류를 의미하는지는 데이터 생성 기준을 확인한 후 판단해야 합니다.

**pandas 확인 목적:**
문자열 날짜를 날짜형으로 변환하면서 변환 실패 값을 찾고, 변환에 성공한 값의 최소·최대 날짜와 시간적 순서를 검증합니다.

## 9. 파일 간 참조 무결성

| 외래 키 관계                  | 참조 대상                   | 참조되지 않는 값 수 | 판정     |
| ------------------------ | ----------------------- | ----------: | ------ |
| `orders.customer_id`     | `customers.customer_id` |           1 | 불일치 존재 |
| `order_items.order_id`   | `orders.order_id`       |           1 | 불일치 존재 |
| `order_items.product_id` | `products.product_id`   |           1 | 불일치 존재 |

세 외래 키 관계에서 **각각 참조 불일치가 1건씩 확인되었습니다.**

다만 이 수치는 관계별 불일치 건수입니다. 실제 고유 문제 행이 정확히 3개라고 단정할 수는 없습니다.

예를 들어 하나의 `order_items` 행에서 `order_id`와 `product_id`가 동시에 일치하지 않는다면 관계 위반은 2건이지만 실제 문제 행은 1행일 수 있습니다. 따라서 원본 데이터 확인 전에는 다음과 같이 구분해야 합니다.

* 관계별 불일치: 각각 1건
* 실제 고유 문제 행 수: 현재 요약만으로 확인 불가

**pandas 확인 목적:**
각 외래 키를 참조 테이블의 기본 키와 비교하여 불일치 행을 추출하고, 여러 관계의 불일치가 동일한 행에서 발생했는지 확인합니다.

## 10. 발견한 문제

### 확인된 문제

1. `orders.customer_id`에서 고객 데이터에 존재하지 않는 참조가 1건 있습니다.
2. `order_items.order_id`에서 주문 데이터에 존재하지 않는 참조가 1건 있습니다.
3. `order_items.product_id`에서 상품 데이터에 존재하지 않는 참조가 1건 있습니다.

### 추가 확인이 필요한 사항

1. `signup_date`와 `order_date`가 문자열로 저장되어 있어 날짜 변환 가능 여부를 확인해야 합니다.
2. 숫자형 분석 컬럼의 실제 범위가 제공되지 않아 음수, 0, 극단값 여부를 확인해야 합니다.
3. 범주형 컬럼의 고유값이 제공되지 않아 오탈자나 표기 혼재 여부를 확인해야 합니다.
4. `products.price`와 `order_items.unit_price`의 차이가 어느 정도 발생하는지 확인해야 합니다.
5. `orders`에는 존재하지만 `order_items`가 연결되지 않는 주문이 있는지 반대 방향 관계도 확인해야 합니다.
6. 주문 상세에서 서로 다른 `order_item_id`가 동일한 주문·상품·수량·단가 조합을 가지는지 확인해야 합니다.

현재 기본 키 결측, 기본 키 중복, 전체 행 중복, 일반 결측치는 발견되지 않았습니다.

## 11. 즉시 처리하지 않은 이유

현재 발견된 외래 키 불일치 행은 원인을 확인하지 않고 즉시 삭제해서는 안 됩니다.

가능한 원인으로는 다음과 같은 가설을 검토할 수 있습니다.

* 데이터셋별 추출 시점이 다름
* 기준 데이터 일부가 누락됨
* 과거 기록이 현재 기준 데이터에 남아 있지 않음
* ID 입력 또는 적재 과정에서 오류가 발생함
* 테스트 또는 별도 관리 대상 데이터가 포함됨

이러한 가능성은 현재 확인된 사실이 아니라 원인 조사를 위한 가설입니다.

또한 다음 값들도 자동으로 삭제하거나 수정해서는 안 됩니다.

* 반복되는 외래 키
* 동일한 고객 이름
* 동일한 상품 이름
* 0원 가격 또는 단가
* 동일한 주문·상품 조합
* `products.price`와 다른 `order_items.unit_price`

각 값이 실제 오류인지 정상적인 업무 기록인지 확인할 수 없기 때문입니다. 따라서 우선 삭제하지 않고 검토 대상으로 분리하는 것이 안전합니다.

## 12. 다음 전처리 단계에서 결정할 내용

### 1순위: 참조 무결성 처리 기준

* 관계별 불일치 행을 실제로 확인합니다.
* 여러 관계의 문제가 동일한 행에서 발생했는지 확인합니다.
* ID 오타인지, 기준정보 누락인지, 추출 시점 차이인지 분류합니다.
* 수정, 보완, 제외 또는 별도 보관 중 처리 방식을 결정합니다.

### 2순위: 날짜 변환 기준

* `signup_date`와 `order_date`를 날짜형으로 변환합니다.
* 변환 실패 값을 별도로 관리할지 결정합니다.
* 분석 기준 기간과 날짜 범위를 결정합니다.
* 가입일과 주문일의 시간 순서 검증 기준을 정합니다.

### 3순위: 숫자형 유효 범위

* `age`, `price`, `quantity`, `unit_price`의 허용 범위를 정합니다.
* 0과 음수를 오류로 볼지 업무 규칙을 확인합니다.
* 극단값을 제거할지, 유지할지, 별도 표시할지 결정합니다.

### 4순위: 범주형 표준화

* `gender`, `city`, `category`, `payment_method`, `order_status`의 고유값을 확인합니다.
* 앞뒤 공백과 표기 차이를 정리합니다.
* 비슷한 값을 하나로 통합할지 실제 업무 의미를 확인한 후 결정합니다.

### 5순위: 가격 컬럼 사용 기준

* `products.price`와 `order_items.unit_price`의 차이를 확인합니다.
* 상품 기준 가격과 주문 당시 단가 중 어떤 값을 분석에 사용할지 결정합니다.
* 주문 금액 계산 시 `quantity`와 `unit_price`를 사용할지 검토합니다.

### 6순위: 조인과 집계 기준

* 고객–주문은 고객 1명 대 주문 여러 건으로 연결합니다.
* 주문–주문 상세는 주문 1건 대 상세 여러 건으로 연결합니다.
* 상품–주문 상세는 상품 1개 대 상세 여러 건으로 연결합니다.
* 병합 후 행 수가 증가하더라도 이를 주문 수 증가로 잘못 해석하지 않도록 합니다.
* 주문 건수는 필요에 따라 고유 `order_id`를 기준으로 집계합니다.
